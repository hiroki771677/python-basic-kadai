```python
class Human:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def check_adult(self):
        if self.age >= 20:
            print(f"{self.name}（{self.age}歳）は大人です。")
        else:
            print(f"{self.name}（{self.age}歳）は大人ではありません。")

# インスタンスを複数生成してリストに追加
humans = [
    Human("太郎", 25),
    Human("花子", 18),
    Human("次郎", 20),
    Human("三郎", 15)
]

# リストの要素数分だけ check_adult メソッドを呼び出す
for human in humans:
    human.check_adult()
```
